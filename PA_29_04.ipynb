```python
# Escrever um programa que o usuário digita uma palavra e o programa  verifica se a palavra é um palíndromo e imprime

palavra = str(input("Digite sua palavra: "))
invertida = ""

for i in (palavra):
  invertida = i + invertida

  print (invertida)



if (invertida) == palavra:
  print (f"Essa palavra é um palíndromo.")


else:
    print (f"Essa palavra não é um palíndromo.")
```

```python
# Peça um número ao usuário e o programa irá separar números pares e ímpares
# Exemplo de saída:

numero = int(input("Digite um número: "))

for i in range (0, numero):
  if i % 2 == 0:
    print (i)

for r in range (numero):
  if r % 3 == 0:
    print (r)
```

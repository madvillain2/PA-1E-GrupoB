```python
# Fazer um código que identifica se um número é par or ímpar

# Entrada

numero = int (input("Digite um número: "))

# Desenvolvimento

par = numero % 2

if par == 0:
  print (f"Seu número é par.")

else:
  print(f"Seu número é ímpar.")
```

```python
# Escrever um programa que receba 04 notas do aluno, calcula a média e mostra se i aluno foi aprovado.


# Entrada

print("Digite suas notas AQUI, valores de 1 à 10 apenas.")
nota1 =  float (input("Digite a nota do primeiro bimestre: "))

if nota1 > 10:
  print(f"Isso não é possível. Digite notas de 0 à 10.")

nota2 =  float (input("DIgite a nota do segundo bimestre: "))

if nota2 > 10:
  print(f"Isso não é possível. Digite notas de 0 à 10.")

nota3 =  float (input("Digite a nota do terceiro bimestre: "))

if nota3 > 10:
  print(f"Isso não é possível. Digite notas de 0 à 10.")

nota4 =  float (input("Digite a nota do quarto bimestre: "))

if nota4 > 10:
  print(f"Isso não é possível. Digite notas de 0 à 10.")

nome = str (input("Digite seu nome: "))

# Processamento

resultado = (nota1 + nota2 + nota3 + nota4) / 4

# Saída

if resultado > 10:
  print(f"{nome}, isso não é possível. Digite notas de 0 à 10.")

elif resultado >= 7:
  print(f"{nome}, sua média é de {resultado}. Você foi aprovado.")

elif resultado >= 5:
  print(f"{nome}, sua média é de {resultado}. Você está de recuperação.")

else:
  print(f"{nome}, sua média é de {resultado}. Você foi reprovado.")
```

```python
# Escrever um programa que recebe os valores de 03 lados e verificar se é um triângulo válido.

# Classificar tipos de triângulos

# A soma de quaisquer dois lados deve ser maior que o terceiro, maior, lado.

# Uma forma prática é garantir que o maior lado seja menor que a soma dos outros dois.

# Entrada

lado1 = float (input("Digite o valor do primeiro lado do triângulo: "))
lado2 = float (input("Digite o valor do segundo lado do triângulo: "))
lado3 = float (input("Digite o valor do terceiro, maior, lado do triângulo: "))

# Processamento

soma = lado1 + lado2

if soma > lado3:
  print(f"Seu triângulo é válido.")

else:
  print(f"Seu triângulo não é válido.")


if lado1 == lado2 and lado1 == lado3:
  print(f"Seu triângulo é equilátero.")


elif lado1 == lado2 or lado1 == lado3 or lado2 == lado3:
  print(f"Seu triângulo é isósceles.")


else:
  print(f"Seu triângulo é escaleno.")
```

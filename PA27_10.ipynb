```python
# 1)

cor = 1

match cor:
  case 1:
    print("Azul")
  case _:
    print("Outra cor")
```

```python
# 2)

numero = 2

match numero:
  case 1:
    print("Um")
  case 2:
    print("Dois")
  case 3:
    print("Três")
```

```python
# 3)

bebida = 1

match bebida:
  case 1:
    print("Água")
  case 2:
    print("Suco")
  case 3:
    print("Refrigerante")
  case _:
    print("Opção inválida")
```

```python
# 4)

dia = 6

if dia == 6 or dia == 7:

  match dia:
   case 1:
    print("Dia de semana")
   case 2:
    print("Dia de semana")
   case 3:
    print("Dia de semana")
   case 4:
    print("Dia de semana")
   case 5:
    print("Dia de semana")
   case _:
    print("Opção inválida")

else:
    print("Final de semana")
```

```python
# 5)

nota = input(str("Digite sua nota, escolha entre 'A', 'B' e 'C': "))

match nota:
   case "A":
    print("Excelente")
   case "B":
    print("Bom")
   case "C":
    print("Precisa melhorar")
   case _:
    print("Nota inválida")
```

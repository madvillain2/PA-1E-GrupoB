```python
# 5)

idade = 16
nota = 8.5
preco = 45.0

if idade > 18:
  print(f"Possui {idade} anos de idade.")
  print("Possui a idade necessária.")
  print("")

else:
  print(f"Possui {idade} anos de idade.")
  print("Não possui a idade necessária.")
  print("")


if nota > 7:
  print(f"Possui {nota} como nota.")
  print("Possui a nota necessária.")
  print("")

else:
  print(f"Possui {nota} como nota.")
  print("Não possui a nota necessária.")
  print("")


if preco < 50:
  print(f"O preço é de {preco}")
  print("Possui a quantia necessária.")
  print("")

else:
  print(f"O preço é de {preco}")
  print("Não possui a quantia necessária.")
  print("")


if idade == 16:
  print(f"Possui {idade} anos de idade.")
  print("Possui a idade igual.")
  print("")

else:
  print(f"Possui {idade} anos de idade.")
  print("Não possui a mesma idade.")
  print("")


if nota != 10:
  print(f"Possui {nota} como nota.")
  print("A nota é diferente de 10.")
  print("")


else:
  print(f"Possui {nota} como nota.")
  print("A nota não é diferente de 10.")
  print("")

```

```python
# 6)

valor = 80
desconto = valor * 0.10
produto_final = valor - desconto

if produto_final < 75:
  print("O valor final é menor que R$75,00.")
  print(f"O valor final é de {produto_final}.")

else:
  print("O valor final é maior que R$75,00.")
  print(f"O valor final é de {produto_final}.")
```

```python
# 1 - Contagem simples, o usuario digita um numero e o programa imprime do zero até o número.

numero = int(input("Digite um numero: "))
for i in range(numero + 1):
  print(i)
```

```python
# 2 - Contagem regressiva, o usuario digita um numero e o programa imprime o número até o zero

numero = int(input("Digite um numero: "))
for i in range(numero , -1, -1):
  print(i)
```

```python
# 3 - Numeros pares, o usuario digita numero e o programa imprime os numeros pares.

i = int(input("Digite um número: "))
for i in range(0, i):
  if i % 2 == 0:
    print(i)
```

```python
# 4 - Numeros impares, o usuario digita um numero e o programa digita os numeros impares

i = int(input("Digite um número: "))
print(1)
for i in range(1, i):
  if i % 3 == 0:
    print(i)

```

```python
# 5 - Soma, o usuario digita um numero e o programa soma os numeros do zero até o numero.

numero = int(input("Digite um número: "))
for i in range(numero):
  print(i + numero)
```

```python
# 6 - Tabuada, o usuario digita um numero e o programa imprime a tabuada do numero

numero = int(input("Digite um número: "))
for i in range(10 + 1):
  print(f" {numero} * {i} = ", numero * i)

```

```python
# 4)

tempo = 0

nota = 0

while tempo != 1:

  nota = int(input("Digite uma nota: "))

  if nota >= 7:
    print(f"Nota {nota}. Aluno aprovado.")

  elif nota >= 5 and nota < 7:
    print(f"Nota {nota}. Aluno em recuperação. ")

  else:
    print(f"Reprovado.")
```

```python
# 6)

tempo = 0


numero = 0


while tempo != 1:
  numero = int(input("Digite um número: "))

  if numero > 0:
    print(f"{numero} é positivo.")
    print("")

  elif numero < 0:
    print(f"{numero} é negativo.")
    print("")

  elif numero == 0:
    print(f"{numero} é igual a zero.")
    print("")

    tempo = int(input("Digite 1 para parar o programa."))
    print("")
```

```python
# 7)

tempo = 0

idade = 0

while tempo != 1:

  idade = int(input("Digite uma idade: "))

  if idade < 12:
    print(f"Menores que 12 anos são crianças.")
    print("")

  elif idade >= 12 and idade <= 17:
    print(f"Maiores que 12 anos porém menores que dezessete anos são adolescentes.")
    print("")

  elif idade >= 18 and idade <= 59:
    print(f"Maiores que 17 anos porém menores que cinquenta e nove anos são adultos.")

  elif idade >= 60:
    print(f"Maiores que 60 anos são idosos.")

    tempo = int(input("Digite 1 para parar o programa."))
    print("")
```

```python
# 8)

tempo = 0

nota = 0
nota2 = 0
media = 0

while tempo != 2:

    nota = float(input("Digite uma nota: "))
    nota2 = float(input("Digite outra nota: "))
    print("")
    media = (nota + nota2) / 2

    if media > 7:
      print(f"A média é de {media}. Aluno aprovado.")
      print("")

    else:
      print(f"Aluno reprovado. Média de {media}.")

    tempo = int(input("Digite 2 para parar o programa: "))
    print("")



```

```python
# 10)

peso = 0

altura = 0

tempo = 0

imc = 0

while tempo != 1:

  altura = float(input("Digite uma altura: "))
  peso = float(input("Digite um peso: "))

  imc = peso / (altura * altura)

  print(f"O IMC é de {imc}")

  tempo = int(input("Digite 1 para parar o programa."))
```

```python
# Peça um número ao usuário e informe se ele é par ou ímpar.


#  Entrada

numero1 = int (input("Digite seu número: "))

# Desenvolvimento

parouimpar = numero1 % 2

# Saída

if parouimpar == 0:
  print (f"Seu número é par.")


else:
  print (f"Seu número é ímpar.")





```

```python
# Maior de dois números
# Solicite dois números e mostre qual é o maior (ou se são iguais).

# Entrada

numero1 = float (input("Digite seu número: "))
numero2 = float (input("Digite seu número: "))

# Desenvolvimento e Saida

if numero1 > numero2:
  print (f"{numero1} é maior que {numero2}")

elif numero1 < numero2:
  print (f"{numero2} é maior que {numero1}")

else:
  print (f"Ambos números são iguais.")
```

```python
# Peça a nota de um aluno e informe:
# Aprovado (nota ≥ 7)
# Recuperação (5 ≤ nota < 7)
# Reprovado (nota < 5)

# Entrada

numero1 = float (input("Digite seu número: "))
numero2 = float (input("Digite seu número: "))
numero3 = float (input("Digite seu número: "))
numero4 = float (input("Digite seu número: "))

# Desenvolvimento

nota_final = (numero1 + numero2 + numero3 + numero4) / 4

# Saída

if nota_final >= 7:
  print (f"Sua nota é de {nota_final}. Você foi aprovado.")

elif nota_final >= 5:
  print (f"Sua nota é de {nota_final}. Você está de recuperação.")

else:
  print (f"Sua nota é de {nota_final}. Você foi reprovado.")
```

```python
# Receba a idade e classifique:
# Criança (0–12)
# Adolescente (13–17)
# Adulto (18–59)
# Idoso (60+)

# Entrada

idade = int (input("Digite sua idade: "))

# Desenvolvimento

if idade <= 12:
  print (f"Sua idade é de uma criança.")

elif idade >= 13 and idade <= 17:
  print (f"Sua idade é de um adolescente.")

elif idade >= 18 and idade <= 59:
  print (f"Sua idade é de um adulto.")

else:
  print (f"Sua idade é de um idoso.")
```

```python
# Peça um número e diga se ele é positivo, negativo ou zero.

numero1 = float (input("Digite seu número: "))

# Desenvolvimento e Saída

if numero1 < 0:
  print (f"Seu número é negativo.")

elif numero1 == 0:
  print (f"Seu número é zero.")

else:
  print (f"Seu número é positivo.")
```

```python
# Peça dois números e uma operação (+, -, *, /).
# Use if/elif para calcular o resultado.

# Entrada

numero1 = float (input("Digite seu número: "))
numero2 = float (input("Digite seu número: "))

operaçao = str (input("Digite qual operação você quer. Entre SOMA, SUBTRAÇÃO, MULTIPLICAÇÃO ou DIVISÃO: "))

# Desenvolvimento

if operaçao == "SOMA":
  print (numero1 + numero2)

elif operaçao == "SUBTRAÇÃO":
  print (numero1 - numero2)

elif operaçao == "MULTIPLICAÇÃO":
  print (numero1 * numero2)

elif operaçao == "DIVISÃO":
  print (numero1 / numero2)
```

```python
# Peça um número de 1 a 7 e mostre o dia correspondente (ex: 1 = domingo).

# Entrada

numero1 = int (input("Digite um número de 1 à 7: "))

# Desenvolvimento e Saída

if numero1 == 1:
  print (f"Este número representa domingo.")

elif numero1 == 2:
  print (f"Este número representa segunda.")

elif numero1 == 3:
  print (f"Este número representa terça-feira.")

elif numero1 == 4:
  print (f"Este número representa quarta-feira.")

elif numero1 == 5:
  print (f"Este número representa quinta-feira.")

elif numero1 == 6:
  print (f"Este número representa sexta-feira.")

elif numero1 == 7:
  print (f"Este número representa sábado.")
```

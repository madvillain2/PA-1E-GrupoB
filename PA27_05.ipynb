```python
# Jogo Menu Inteligente

# Criar um sistema que funcione até o usuário escolher sair

# 1 - Par / Ímpar
# 2 - Positivo / Negativo
# 3 - Maior
# 4 - Sair

opcao = 0

while opcao != 4:

  print(f"1 - Par / Ímpar")
  print(f"2 - Positivo / Negativo")
  print(f"3 - Maior")
  print(f"Sair")

  opcao = int(input("Digite a opção desejada: "))

  if opcao == 1:
    parimpar = int(input(f"Digite um número:"))
    if parimpar % 2 == 0:
      print(f"Seu número é par.")

    else:
      print(f"Seu número é ímpar.")

  if opcao == 2:
    positivonegativo = int(input(f"Digite um número:"))
    if 0 > positivonegativo:
      print(f"Seu número é negativo.")


    elif positivonegativo > 0:
      print(f"Seu número é positivo.")



  if opcao == 3:
    maiormenor1 = float(input(f"Digite um número:"))
    maiormenor2 = float(input(f"Digite outro número: "))
    if maiormenor1 > maiormenor2:
      print(f" {maiormenor1} é maior que {maiormenor2}")

    elif maiormenor1 < maiormenor2:
      print(f" {maiormenor2} é maior que {maiormenor1}")

  if opcao == 4:
    break









```

```python
# Urna Eletrônica

# Criar um programa de votação com três candidatos

# Votar candidato A
# Votar candidato B
# Votar candidato C
# Votar Branco
# Votar Nulo
# Encerrar votação

# O sistema deve repetir até encerrar a votação
# Contador votos de cada candidato
# Contar votos brancos e nulos
# Ao encerrar, mostrar:

# Total de Votos
# Total de votos por candidato
# Total votos brancos
# Total de votos nulos
# Vencedor da eleição

# Regras extras

# Se houver empate entre candidatos: Mostrar eleição empatada
# Se não houver votos válidos, mostrar: "nenhum candidato recebeu votos."

opcao = 0
aeme = 0
ye = 0
stalling = 0
branco = 0
nulo = 0



while opcao != 6:

  print(f"1 - Votar em AM.")
  print(f"2 - Votar em Kanye West.")
  print(f"3 - Votar em Stalin")
  print(f"4 - Votar branco")
  print(f"5 - Votar nulo")
  print(f"6 - Encerrar")

  opcao = int(input(f"Escolha sua opção: "))

  if opcao == 1:
    aemevoto = int(input("Você quer votar em AM? Digite '1' para confirmar."))
    if aemevoto == 1:
      aeme += 1
      print(f"Você votou em AM.")

  if opcao == 2:
      yevoto = int(input("Você quer votar em Kanye West? Digite '1' para confirmar."))
      if yevoto == 1:
        ye += 1
        print(f"Você votou em Kanye West.")

  if opcao == 3:
        stallingvoto = int(input("Você quer votar em Josef Stalin? Digite '1' para confirmar."))
        if stallingvoto == 1:
          stalling += 1
          print(f"Você votou em Josef Stalin.")

  if opcao == 4:
        brancovoto = int(input(f"Você quer votar branco? Digite '1' para confirmar."))
        if brancovoto == 1:
          branco += 1
          print(f"Você votou branco.")

  if opcao == 5:
        nulovoto = int(input(f"Você quer votar nulo? Digite '1' para confirmar."))
        if nulovoto == 1:
          nulo += 1
          print(f"Você votou nulo.")

  if opcao == 6:
    print(f"AM ficou com {aeme} votos.")
    print(f"Kanye West ficou com {ye} votos.")
    print(f"Josef Stalin ficou com {stalling} votos.")
    print(f"{branco} votos brancos.")
    print(f"{nulo} votos nulos.")


    if aeme > ye and aeme > stalling:
      print(f"AM ganhou a eleição!")

    if ye > aeme and ye > stalling:
      print(f"Kanye Omari West ganhou a eleição!")

    if stalling > aeme and stalling > ye:
      print(f"Josef Stalin ganhou a eleição.")

    if aeme == ye or aeme == stalling or ye == stalling:
      print(f"Eleição empatada!")

    if aeme == 0 and ye == 0 and stalling == 0:
      print(f"Nenhum candidato recebeu votos.")




```

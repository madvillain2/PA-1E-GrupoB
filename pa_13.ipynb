```python
# 9)

matriz = [
[12, 5, 8],
[20, 3, 15],
[7, 25, 10]
]

maior = matriz[0][0]

for linha in matriz:
  for valor in linha:
    if valor > maior:
      maior = valor

print(f"Maior valor: {maior}")
```
